## Step 1: Import Libraries

**Philosophy:** Think of libraries as toolboxes. We don't build our own hammer — we import one.
- `pandas` → the main tool for working with tabular data (like Excel, but in Python)
- `os` → lets Python talk to your file system (find folders, files)


```python
import pandas as pd   # pd is just a short nickname for pandas — everyone uses this
import os             # os = operating system tools

print('✅ Libraries loaded successfully!')
```

## Step 2: Set File Paths

**Philosophy:** Instead of hardcoding file names everywhere, we define them once here.
If you rename a file, you only need to change it in one place.

```python
# os.path.join builds a file path that works on Windows, Mac, and Linux
# '..' means 'go one folder up' from the notebooks/ folder

IBM_PATH = os.path.join('..', 'data', 'raw', 'ibm_hr_attrition.csv')
LINKEDIN_PATH = os.path.join('..', 'data', 'raw', 'linkedin_jobs.csv')

print('IBM path:', IBM_PATH)
print('LinkedIn path:', LINKEDIN_PATH)
```

## Step 3: Load Dataset 1 — IBM HR Attrition

**Philosophy:** `pd.read_csv()` reads a CSV file and converts it into a **DataFrame**.
A DataFrame is like an Excel spreadsheet inside Python — rows and columns you can analyze.

```python
ibm_df = pd.read_csv(IBM_PATH)

print('✅ IBM dataset loaded!')
print(f'Shape: {ibm_df.shape[0]} rows × {ibm_df.shape[1]} columns')
```

```python
# .head() shows the first 5 rows — a quick sanity check
# This is always the first thing you do after loading data
ibm_df.head()
```

```python
# .info() gives a summary: column names, data types, and non-null counts
# If a column shows fewer non-null values than total rows → missing data!
ibm_df.info()
```

```python
# Check how many employees LEFT vs STAYED
# This is your TARGET VARIABLE — what you want to predict
print('Attrition (target variable) breakdown:')
print(ibm_df['Attrition'].value_counts())
print()
print('As percentages:')
print(ibm_df['Attrition'].value_counts(normalize=True).round(3) * 100, '%')
```

```python
# .describe() gives basic statistics for all numeric columns
# min, max, mean, standard deviation — a first feel for the data range
ibm_df.describe().round(2)
```

## Step 4: Load Dataset 2 — LinkedIn Job Postings

This dataset has job listings from 2024. We'll use it to extract:
- **Salary benchmarks by job title** → compare against IBM employees' salaries
- **Job demand by role** → is there a hot job market for a specific role? Does that increase attrition?
- **Remote work availability** → another external pull factor

```python
linkedin_df = pd.read_csv(LINKEDIN_PATH)

print('✅ LinkedIn dataset loaded!')
print(f'Shape: {linkedin_df.shape[0]} rows × {linkedin_df.shape[1]} columns')
```

```python
linkedin_df.head()
```

```python
linkedin_df.info()
```

```python
# See what columns exist in the LinkedIn dataset
print('LinkedIn Columns:')
for col in linkedin_df.columns:
    print(f'  - {col}')
```

## Step 5: Quick Missing Data Check

**Philosophy:** Missing data is the enemy. You need to know *where* and *how much* before you proceed.
A column with 70%+ missing is usually dropped. One with <5% missing can be filled (imputed).

```python
print('=== IBM Dataset: Missing Values ===')
ibm_missing = ibm_df.isnull().sum()
print(ibm_missing[ibm_missing > 0] if ibm_missing.sum() > 0 else '✅ No missing values!')

print()

print('=== LinkedIn Dataset: Missing Values ===')
linkedin_missing = linkedin_df.isnull().sum()
linkedin_missing_pct = (linkedin_missing / len(linkedin_df) * 100).round(1)
missing_report = pd.DataFrame({'Missing Count': linkedin_missing, 'Missing %': linkedin_missing_pct})
print(missing_report[missing_report['Missing Count'] > 0].sort_values('Missing %', ascending=False))
```

## Step 6: Save a Summary Report

```python
summary = f"""
=== DATA LOADING SUMMARY ===

IBM HR Attrition Dataset:
  Rows    : {ibm_df.shape[0]}
  Columns : {ibm_df.shape[1]}
  Target  : Attrition — Yes({ibm_df['Attrition'].value_counts()['Yes']}) / No({ibm_df['Attrition'].value_counts()['No']})
  Missing : {ibm_df.isnull().sum().sum()} cells

LinkedIn Job Postings Dataset:
  Rows    : {linkedin_df.shape[0]}
  Columns : {linkedin_df.shape[1]}
  Missing : {linkedin_df.isnull().sum().sum()} cells
"""

print(summary)

# Save it
with open('../reports/01_data_loading_summary.txt', 'w') as f:
    f.write(summary)

print('✅ Summary saved to reports/01_data_loading_summary.txt')
```
